```python
import urllib.request

base = '/content/drive/MyDrive/AI_Project_Body_Fat_Percentage_Mesuring'
raw = f'{base}/Raw_Data'

urllib.request.urlretrieve(
    "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2005/DataFiles/DEMO_D.xpt",
    f'{raw}/DEMO_D.XPT'
)
urllib.request.urlretrieve(
    "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2005/DataFiles/BMX_D.xpt",
    f'{raw}/BMX_D.XPT'
)

print(os.listdir(raw))
```

```python
import pandas as pd

raw = f'{base}/Raw_Data'

demo = pd.read_sas(f'{raw}/DEMO_D.XPT', format='xport')
bmx = pd.read_sas(f'{raw}/BMX_D.XPT', format='xport')
dxa = pd.read_sas(f'{raw}/DXX_D.XPT', format='xport')

print("DEMO:", demo.shape)
print("BMX:", bmx.shape)
print("DXA:", dxa.shape)
```

```python
print(list(dxa.columns))
```

```python
dxa_valid = dxa[dxa['DXAEXSTS'] == 1]
dxa_clean = dxa_valid.groupby('SEQN').mean().reset_index()
print(dxa_clean.shape)
```

```python
merged = dxa_clean.merge(demo, on='SEQN', how='inner').merge(bmx, on='SEQN', how='inner')
print(merged.shape)
```

```python
print(list(merged.columns))
```

```python
selected_columns = [
    'SEQN',
    'RIDAGEYR', 'RIAGENDR',
    'BMXWT', 'BMXHT', 'BMXBMI', 'BMXWAIST', 'BMXARMC', 'BMXCALF', 'BMXTHICR',
    'DXDTOPF'
]

final_df = merged[selected_columns]
print(final_df.shape)
final_df.head()
```

```python
print(final_df.isna().sum())
```

```python
final_clean = final_df.dropna()
print(final_clean.shape)
```

```python
print(final_clean.describe())
```

```python
final_clean.to_csv(f'{base}/Processed/clean_bodyfat_dataset.csv', index=False)
print("Saved:", f'{base}/Processed/clean_bodyfat_dataset.csv')
```
